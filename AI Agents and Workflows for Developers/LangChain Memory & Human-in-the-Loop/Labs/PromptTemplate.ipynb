```python
from langchain.messages import HumanMessage, SystemMessage
from langchain_core.prompts import ChatPromptTemplate, PromptTemplate
from langchain_core.prompts.chat import AIMessagePromptTemplate, HumanMessagePromptTemplate, MessagesPlaceholder
```

## с PromptTemplate & ChatPromptTemplate можем да организираме промптовете с които работим

има и при работата с reatrieveri използва PromptTemplate- имаме нужда да форматираме документите, които retrievera e открил като съвпадение на query-то изпратено от агента

## PromptTemplate: чрез него получаваме стринг с който да създадем system,human, tool message

```python
prompt_template = PromptTemplate.from_template("Given the user's role ({role}), answer this question:\n{question}") # дефинираме шаблон за нашият промпт
```

```python
prompt_template
```

```python
prompt_template.format(role = "admin", question = "How do I reset my n8n password?")
```

## ChatPromptTemplate този шаблон ни дава въможност да изконфигурираме цял списък от съобщения
  - в MessagesPlaceholder мога да закача множество допълнителни съобщения

```python
chat_prompt_template = ChatPromptTemplate.from_messages(
    messages = [
        SystemMessage("You are a helpful AI assistant."),
        MessagesPlaceholder(variable_name="history", optional=True),
        HumanMessagePromptTemplate.from_template("I am working as a {job_position} in \"{company}\". What is the average salary for my position in the other companies?"),
        AIMessagePromptTemplate.from_template("The average salary for the \"{job_position}\". in other companies is 500$."),
        HumanMessage("What other carreer alternatives do I have nowadays?")
    ]
)
```

```python
chat_prompt_template.format_messages(job_position ="cashier", company="Local SuperMarket")
```

```python
chat_prompt_template.format_messages(job_position ="cashier", company="Local SuperMarket", history = [HumanMessage("I'm from Bulgaria. I am 24-years-old.")])
```
