```python
from google.colab import userdata
from langchain_community.document_loaders import GithubFileLoader, GitHubIssuesLoader, NotionDBLoader
```

```python
github_access_token=userdata.get("GITHUB_TOKEN")
github_file_loader = GithubFileLoader(
    repo="TonyTroeff/Training-SoftUni",
    access_token = github_access_token,
    file_filter=lambda file_path: file_path.find('#') == -1 and file_path.endswith(".txt"))

github_issues_loader = GitHubIssuesLoader(
    repo = "TryAtSoftware/CleanTests",
    access_token= github_access_token
)

notion_access_token = userdata.get('NOTION_TOKEN2')
notion_loader = NotionDBLoader(integration_token=notion_access_token, database_id="68cf327f2c324c3886ba4b2656603d2a")
```
